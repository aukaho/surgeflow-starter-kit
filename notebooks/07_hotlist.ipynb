# 07 Hotlist

The Hotlist and session turnover board are each fetched once for `us`. Change the market to `cn`, `jp`, or `hk` when needed.

Only a response marked as session-live describes the current session. Other responses can be delayed or historical.

Research data only, not investment advice. Each panel refreshes on a schedule; each panel shows its source date.

Free access grants no redistribution rights. **Do not republish the data you fetch.**

```python
import sys
from pathlib import Path
ROOT = Path.cwd()
while ROOT != ROOT.parent and not (ROOT / 'notebooks' / '_shared.py').exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'notebooks'))
sys.path.insert(0, str(ROOT / 'python' / 'src'))
from _shared import fetch_panel, get_api_key, make_client, nested
client, catalogue = make_client()
api_key = get_api_key()
market = 'us'
```

```python
hotlist_columns = {
    'hotlist_rank': 'The position assigned by the Hotlist response.',
    'market': 'The market attached to the row.',
    'ticker': 'The market identifier for the company.',
    'company_name': 'The company name supplied by the response.',
    'industry': 'The industry label supplied by the response.',
    'factor_style': 'The style summary supplied by the response.',
    'price': 'The source price attached to the response.',
    'intraday_return_pct': 'The intraday percentage change attached to the response.',
    'turnover_per_second': 'The turnover rate attached to the response.',
    'market_cap_usd': 'The company market value in US dollars supplied by the response.',
    'projected_turnover_usd': 'The projected session turnover in US dollars.',
    'projected_vs_yesterday': 'Projected turnover divided by the prior-session comparison.',
    'previous_day_turnover_usd': 'The prior-session turnover in US dollars.',
}
hotlist_frame = fetch_panel('Hotlist board', lambda: client.member_hotlist(market, api_key), lambda p: nested(p, 'data', 'rows') or [], hotlist_columns, source_date=lambda p: nested(p, 'data', 'as_of_utc'), status=lambda p: str(nested(p, 'data', 'market_status')) + '; quality=' + str(nested(p, 'data', 'data_quality')), message=lambda p: nested(p, 'data', 'stale_reason') or (None if nested(p, 'data', 'rows') else 'No Hotlist rows returned.'))
```

```python
realtime_columns = {
    'rank': 'The position assigned by the turnover-board response.',
    'ticker': 'The market identifier for the company.',
    'company_name': 'The company name supplied by the response.',
    'price': 'The source price attached to the response.',
    'intraday_return_pct': 'The intraday percentage change attached to the response.',
    'turnover_per_second': 'The turnover rate attached to the response.',
    'accumulated_turnover': 'The turnover accumulated at the source timestamp.',
    'projected_turnover': 'The projected session turnover.',
    'projected_vs_yesterday': 'Projected turnover divided by the prior-session comparison.',
    'previous_day_turnover': 'The prior-session turnover comparison.',
}
realtime_frame = fetch_panel('Session turnover board', lambda: client.member_realtime(market, api_key), lambda p: nested(p, 'data', 'rows') or [], realtime_columns, source_date=lambda p: nested(p, 'data', 'as_of_utc'), status=lambda p: str(nested(p, 'data', 'market_status')) + '; quality=' + str(nested(p, 'data', 'data_quality')), message=lambda p: nested(p, 'data', 'stale_reason') or (None if nested(p, 'data', 'rows') else 'No turnover rows returned.'))
```
